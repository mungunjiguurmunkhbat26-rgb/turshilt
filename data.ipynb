```python
#uvchilsun humuusiin dundaj nas
dundaj_nas = data.groupby('Outcome')['Age'].mean()
print("Өвчлсөн хүмүүсийн дундаж нас:")
print(dundaj_nas)
```

```python
#2 uvchtei bolon eruul humuusiin glyukoziin hemjeenii deed dood
grouped = data.groupby('Outcome')['Glucose']
max_glucose = grouped.max()
min_glucose = grouped.min()
glucose_table = pd.DataFrame({
    'Max_Glucose': max_glucose,
    'Min_Glucose': min_glucose
})

print("Glucose-ийн хамгийн их ба хамгийн бага утга :")
print(glucose_table)


```

```python
#3 heden nastai humuusiin dund uvchlul heden udaa burtgegdsen bna ve
age_bins = [0, 20, 30, 40, 50, 60, 100]
age_labels = ['0-20','21-30','31-40','41-50','51-60','61+']
data['AgeGroup'] = pd.cut(data['Age'], bins=age_bins, labels=age_labels, right=True)
# 3. Outcome=0 (өвчлөгсөд) ба Outcome=1 (эрүүл) тус бүрт насны ангиллаар бүлэглэж тооцоолох
uvchlul_counts = data.groupby(['Outcome','AgeGroup']).size().reset_index(name='Count')
# groupby(['Outcome','AgeGroup']) → Outcome ба AgeGroup тус бүрээр бүлэглэнэ
# size() → тухайн бүлэгт хэдэн хүн байгаа
# reset_index(name='Count') → үр дүнг датафрейм болгож, Count нэртэй багана үүсгэнэ

# 4. Үр дүнг хэвлэх
print("Насны ангилал тус бүрт бүртгэл (өдөрт хэд дахин бүртгэгдсэнтэй адил):")
print(uvchlul_counts)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv('/content/diabetes.csv')
def corrHeatMap(df):
    _, axes = plt.subplots(figsize=(10, 8))
    clr = sns.diverging_palette(220, 10, as_cmap=True)

    sns.heatmap(
        df.corr(),
        cmap=clr,
        square=True,
        cbar_kws={'shrink': 0.5},
        ax=axes,
        annot=True,
        linewidths=.1,
        vmax=1.0,
        linecolor='white',
        annot_kws={'fontsize': 8}
    )
    plt.title('Пийрсоны корреляци', y=1.05, size=15)
    plt.show()
neededFields = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'
]
corrHeatMap(data[neededFields])
```
