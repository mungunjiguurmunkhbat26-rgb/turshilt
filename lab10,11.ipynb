```python

class Rectangle:
    # 1. Байгуулагч функц
    def __init__(self, a=None, b=None, name="Rectangle", copy=None):
        if copy is not None:
            # Өөр объектоос хуулж авах
            self.a = copy.a
            self.b = copy.b
            self.name = copy.name

        elif a is not None and b is not None:
            # Параметрээс утга авах
            self.a = a
            self.b = b
            self.name = name

        else:
            # Гараас утга унших
            self.a = float(input("Уртыг оруулна уу: "))
            self.b = float(input("Өргөнийг оруулна уу: "))
            self.name = input("Нэрийг оруулна уу: ")

    # 2. Талбай олох функц
    def area(self):
        return self.a * self.b

    # 3. Периметр олох функц
    def perimeter(self):
        return 2 * (self.a + self.b)

    # 4. Мэдээлэл хэвлэх функц
    def display(self):
        print(f"\nНэр: {self.name}")
        print(f"
        Урт: {self.a}")
        print(f"Өргөн: {self.b}")
        print(f"Талбай: {self.area()}")
        print(f"Периметр: {self.perimeter()}")


# 1-р объект: Гараас утга авах
r1 = Rectangle()

# 2-р объект: Параметрээс утга авах
r2 = Rectangle(5, 3, "Тэгш өнцөгт 2")

# 3-р объект: Файлаас утга авах
with open("rectangle.txt", "w", encoding="utf-8") as f:
    f.write("8\n4\nТэгш өнцөгт 3\n")

with open("rectangle.txt", "r", encoding="utf-8") as f:
    a = float(f.readline())
    b = float(f.readline())
    name = f.readline().strip()

r3 = Rectangle(a, b, name)

# 4-р объект: Өөр объектоос хуулж авах
r4 = Rectangle(copy=r2)
r4.name = "Тэгш өнцөгт 4 (хуулбар)"

# Үр дүнг хэвлэх
r1.display()
r2.display()
r3.display()
r4.display()
```

```python

import math


class EqQuad:
    # Байгуулагч функц
    def __init__(self, a=None, b=None, c=None, copy=None):

        if copy is not None:
            # 1. Өөр объектоос хуулж авах
            self.a = copy.a
            self.b = copy.b
            self.c = copy.c

        elif a is not None and b is not None and c is not None:
            # 2. Параметрээс утга авах
            self.a = a
            self.b = b
            self.c = c

        else:
            # 3. Гараас утга авах
            self.a = float(input("a коэффициентийг оруул: "))
            self.b = float(input("b коэффициентийг оруул: "))
            self.c = float(input("c коэффициентийг оруул: "))

    # Квадрат тэгшитгэлийн шийд олох функц
    def bodolt(self):
        print(f"\nТэгшитгэл: {self.a}x^2 + "
              f"{self.b}x + {self.c} = 0")

        if self.a == 0:
            if self.b != 0:
                x = -self.c / self.b
                print("Квадрат биш, шугаман тэгшитгэл.")
                print("Шийд: x =", x)
            elif self.c == 0:
                print("Хязгааргүй олон шийдтэй.")
            else:
                print("Шийдгүй.")
            return

        D = self.b ** 2 - 4 * self.a * self.c

        print("Дискриминант D =", D)

        if D > 0:
            x1 = (-self.b + math.sqrt(D)) / (2 * self.a)
            x2 = (-self.b - math.sqrt(D)) / (2 * self.a)
            print("Хоёр ялгаатай бодит шийдтэй.")
            print("x1 =", x1)
            print("x2 =", x2)

        elif D == 0:
            x = -self.b / (2 * self.a)
            print("Нэг давхар бодит шийдтэй.")
            print("x =", x)

        else:
            print("Бодит шийдгүй.")


# 1-р объект: Гараас утга авах
print("1. Гараас утга авах")
eq1 = EqQuad()
eq1.bodolt()


# 2-р объект: Параметрээс утга авах
print("\n2. Параметрээс утга авах")
eq2 = EqQuad(1, -5, 6)
eq2.bodolt()


# 3-р объект: Файлаас утга авах
print("\n3. Файлаас утга авах")

with open("equation.txt", "w", encoding="utf-8") as f:
    f.write("1\n-3\n2\n")

with open("equation.txt", "r", encoding="utf-8") as f:
    a = float(f.readline())
    b = float(f.readline())
    c = float(f.readline())

eq3 = EqQuad(a, b, c)
eq3.bodolt()


# 4-р объект: Өөр объектоос хуулж авах
print("\n4. Өөр объектоос хуулж авах")
eq4 = EqQuad(copy=eq2)
eq4.bodolt()
```
